####Problema 1. Suma de vectores
---

El senderismo, considerado un deporte de aventura, consiste en realizar recorridos a pie a través de espacios naturales. Supón que un senderista sigue un itinerario que consta de tres desplazamientos: el primero, de 5 km, 60° al norte del este; el segundo, de 2 km, al norte; y el último, de 3 km, 30° al sur del oeste.

a) 	Calcula las componentes $x$ y $y$ de cada desplazamiento.

>**Solución:**

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Configuración de los Vectores ---

# Vector 1 (V1): 5 km, 60° al norte del este
# Ángulo estándar = 60°
mag1 = 5.0
ang1_deg = 60.0
ang1_rad = np.deg2rad(ang1_deg) # Se convierten los ángulos a radianes para que las funciones de numpy (cos, sin) funcionen

# Vector 2 (V2): 2 km, al norte
# Ángulo estándar = 90°
mag2 = 2.0
ang2_deg = 90.0
ang2_rad = np.deg2rad(ang2_deg)

# Vector 3 (V3): 3 km, 30° al sur del oeste
# "Oeste" es 180°. "30° al sur" significa que le sumamos 30° más.
# Ángulo estándar = 180° + 30° = 210°
mag3 = 3.0
ang3_deg = 210.0
ang3_rad = np.deg2rad(ang3_deg)

# --- a) Calcula las componentes x e y de cada desplazamiento. ---
v1_x = mag1 * np.cos(ang1_rad)
v1_y = mag1 * np.sin(ang1_rad)

v2_x = mag2 * np.cos(ang2_rad)
v2_y = mag2 * np.sin(ang2_rad)

v3_x = mag3 * np.cos(ang3_rad)
v3_y = mag3 * np.sin(ang3_rad)

print("--- Apartado (a): Componentes de los Vectores ---")
print(f"Vector 1 (V1): x = {v1_x:.3f} km, y = {v1_y:.3f} km")
print(f"Vector 2 (V2): x = {v2_x:.3f} km, y = {v2_y:.3f} km")
print(f"Vector 3 (V3): x = {v3_x:.3f} km, y = {v3_y:.3f} km")
print("\n")
```

b)	Halla la magnitud y el ángulo del desplazamiento resultante. Expresa la respuesta en kilómetros y en millas.

>**Solución:**

```python
# 1. Sumar componentes
R_x = v1_x + v2_x + v3_x
R_y = v1_y + v2_y + v3_y

# 2. Calcular magnitud (Teorema de Pitágoras)
# R = sqrt(R_x^2 + R_y^2)
magnitud_km = np.sqrt(R_x**2 + R_y**2)

# 3. Calcular ángulo
# Usamos arctan2(y, x) porque maneja correctamente todos los cuadrantes
angulo_rad = np.arctan2(R_y, R_x)
angulo_deg = np.rad2deg(angulo_rad)

# 4. Convertir magnitud a millas
km_a_millas = 0.621371
magnitud_millas = magnitud_km * km_a_millas

print("--- Apartado (b): Vector Resultante ---")
print(f"Componente Resultante: R_x = {R_x:.3f} km, R_y = {R_y:.3f} km")
print(f"Magnitud: {magnitud_km:.3f} km")
print(f"Magnitud: {magnitud_millas:.3f} millas")
print(f"Ángulo: {angulo_deg:.3f}° (medido en contra de las manecillas del reloj desde el Este)")
print("\n")
```

c)	Traza la gráfica de los tres desplazamientos sucesivos y del desplazamiento resultante sobre el mismo plano cartesiano.

>**Solución:**

```python
# --- c) Traza la gráfica de los tres desplazamientos y el resultante. ---
print("--- Apartado (c): Generando Gráfica ---")

# Puntos para la gráfica (cabeza-cola)
p0 = [0, 0]
p1 = [v1_x, v1_y]
p2 = [v1_x + v2_x, v1_y + v2_y]
p3 = [R_x, R_y] # El punto final es el vector resultante

plt.figure(figsize=(10, 8))

# Opciones de 'quiver' (flechas): quiver(X_inicio, Y_inicio, X_componente, Y_componente, ...)
# scale=1, angles='xy', scale_units='xy' -> Dibuja el vector con las unidades del eje

# Vector 1 (desde el origen)
plt.quiver(p0[0], p0[1], v1_x, v1_y, angles='xy', scale_units='xy', scale=1, color='blue', label=f'V1 (5 km, {ang1_deg}°) - ({v1_x:.2f}, {v1_y:.2f})')

# Vector 2 (desde el final de V1)
plt.quiver(p1[0], p1[1], v2_x, v2_y, angles='xy', scale_units='xy', scale=1, color='green', label=f'V2 (2 km, {ang2_deg}°) - ({v2_x:.2f}, {v2_y:.2f})')

# Vector 3 (desde el final de V2)
plt.quiver(p2[0], p2[1], v3_x, v3_y, angles='xy', scale_units='xy', scale=1, color='orange', label=f'V3 (3 km, {ang3_deg}°) - ({v3_x:.2f}, {v3_y:.2f})')

# Vector Resultante (desde el origen hasta el final de V3)
plt.quiver(p0[0], p0[1], R_x, R_y, angles='xy', scale_units='xy', scale=1, color='red', label=f'Resultante (R) - ({R_x:.2f}, {R_y:.2f})')

# --- Configuración de la Gráfica ---
plt.title('Suma de Vectores (Paseo del Senderista)')
plt.xlabel('Desplazamiento Este-Oeste (km)')
plt.ylabel('Desplazamiento Norte-Sur (km)')

# Establecer límites para ver todos los vectores
# Encontramos los puntos intermedios para ajustar bien los ejes
all_x_points = [p0[0], p1[0], p2[0], p3[0]]
all_y_points = [p0[1], p1[1], p2[1], p3[1]]
plt.xlim(min(all_x_points) - 1, max(all_x_points) + 1)
plt.ylim(min(all_y_points) - 1, max(all_y_points) + 1)

plt.axhline(0, color='black', linewidth=0.5) # Eje X
plt.axvline(0, color='black', linewidth=0.5) # Eje Y
plt.grid(True, linestyle='--')
plt.legend()
plt.axis('equal') # ¡MUY IMPORTANTE para que los ángulos se vean correctos!
plt.show()
```

####Problema 2. Movimiento en una dimensión  
---

Un automóvil pasa con una rapidez constante de $30~m/s$ por un puesto de control de la policía de tránsito. Luego de 2 segundos, una patrulla sale en su persecución, con una aceleración constante de $3~m/s^{2}$.

a) 	Escribe las ecuaciones de movimiento del automóvil y de la patrulla de policía.

>**Solución:**

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Datos del Problema ---
v_auto = 30.0   # m/s (velocidad constante del automóvil)
a_patrulla = 3.0  # m/s^2 (aceleración de la patrulla)
retraso = 2.0     # s (tiempo que la patrulla espera)

# --- a) Ecuaciones de movimiento ---
# Usaremos 't' como el tiempo total desde que el automóvil pasa el control (t=0)

# Ecuación del automóvil (MRU): x = v*t
# x_auto(t) = 30t
print("--- Apartado (a): Ecuaciones de Movimiento ---")
print("Ecuación del Automóvil: x_c(t) = 30t")

# Ecuación de la patrulla (MRUA): x = v0*t' + 0.5*a*t'^2
# Comienza 2 segundos después, así que su tiempo de movimiento es (t - 2)
# Su movimiento es válido solo para t >= 2 segundos.
# x_patrulla(t) = 0.5 * 3 * (t - 2)^2
print("Ecuación de la Patrulla: x_p(t) = 1.5 * (t - 2)^2   (para t >= 2 s)")
print("\n")
```

b) 	Utiliza las dos ecuaciones del literal anterior para calcular el tiempo que tarda la patrulla en alcanzar al automóvil.

>**Solución:**

```python

# --- b) Calcular el tiempo de alcance ---
# El alcance ocurre cuando x_auto = x_patrulla
# 30t = 1.5 * (t - 2)^2
# 30t = 1.5 * (t^2 - 4t + 4)
# 30t = 1.5t^2 - 6t + 6
# 0 = 1.5t^2 - 36t + 6

# Resolvemos la ecuación cuadrática ax^2 + bx + c = 0
a = 1.5
b = -36.0
c = 6.0

# Usamos la función 'roots' de numpy para encontrar las raíces de la ecuación
coeficientes = [a, b, c]
soluciones = np.roots(coeficientes)

print("--- Apartado (b): Tiempo de Alcance ---")
print(f"Soluciones matemáticas (raíces) de la ecuación: {soluciones[0]:.3f} s y {soluciones[1]:.3f} s")

# Filtramos la solución físicamente imposible
# La patrulla no se mueve antes de t=2s, así que cualquier solución < 2 es inválida
t_alcance = 0
if soluciones[0] > retraso:
    t_alcance = soluciones[0]
else:
    t_alcance = soluciones[1]

# Calculamos la distancia donde se encuentran
d_alcance = v_auto * t_alcance

print(f"La solución físicamente válida es t = {t_alcance:.3f} segundos.")
print("La otra solución se descarta porque es antes de que la patrulla comience a moverse (t < 2s).")
print(f"La patrulla tarda {t_alcance:.3f} s (desde que el auto pasó el control) en alcanzarlo.")
print(f"El tiempo que la patrulla estuvo en movimiento fue: {t_alcance - retraso:.3f} s")
print(f"Se encuentran a una distancia de {d_alcance:.2f} metros del puesto de control.")
print("\n")
```

c) 	Traza las gráficas de distancia contra tiempo del automóvil y de la patrulla sobre el mismo plano cartesiano. Marca el punto de corte de las gráficas y explica su significado en el contexto del problema.

>**Solución:**

```python
# --- c) Trazar las gráficas ---
print("--- Apartado (c): Gráfica de Distancia vs. Tiempo ---")

# Creamos un rango de tiempo, un poco más allá del tiempo de alcance
t_max = t_alcance + 5  # Graficamos 5 segundos más
tiempo = np.linspace(0, t_max, 500) # Vector de tiempo

# Calculamos la posición del automóvil (MRU)
x_auto = v_auto * tiempo

# Calculamos la posición de la patrulla (MRUA con retraso)
# Usamos np.maximum(0, ...) para que el tiempo (t-2) no sea negativo
# y la patrulla se quede en x=0 antes de t=2
tiempo_patrulla = np.maximum(0, tiempo - retraso)
x_patrulla = 0.5 * a_patrulla * (tiempo_patrulla**2)

# Crear la gráfica
plt.figure(figsize=(10, 6))
plt.plot(tiempo, x_auto, label='Automóvil (Vel. constante)', color='blue')
plt.plot(tiempo, x_patrulla, label='Patrulla (Acelerando)', color='red', linestyle='--')

# Marcar el punto de corte (intersección)
plt.scatter([t_alcance], [d_alcance], color='black', zorder=5, s=100, label=f'Alcance ({t_alcance:.2f} s, {d_alcance:.2f} m)')
plt.text(t_alcance, d_alcance - 100, f'  ({t_alcance:.2f} s, {d_alcance:.2f} m)', horizontalalignment='left')

# Añadir títulos y etiquetas
plt.title('Gráfica de Movimiento: Automóvil vs. Patrulla')
plt.xlabel('Tiempo (segundos)')
plt.ylabel('Distancia (metros)')
plt.legend()
plt.grid(True)
plt.show()
```

####Problema 3. Movimiento en dos dimensiones  
---

Un proyectil se lanza con una rapidez inicial de $v_{0}~m/s$, formando un ángulo de $\theta°$ con la horizontal.

a) Escribe una función que calcule la altura máxima y el alcance horizontal del proyectil, utilizando como datos de entrada la rapidez inicial y el ángulo de disparo.

>**Solución:**

```python
import numpy as np
import pandas as pd

# --- Definir constantes ---
# Asumimos la aceleración de la gravedad estándar en la Tierra
g = 9.8  # m/s^2

# --- Apartado (a): Función para calcular altura y alcance ---
# Esta es la solución para el apartado (a).
# La función toma la velocidad y el ángulo en grados,
# y devuelve la altura máxima y el alcance.

def calcular_proyectil(v0, angulo_grados):
    """
    Calcula la altura máxima y el alcance horizontal de un proyectil.

    Argumentos:
    v0 (float): Rapidez inicial en m/s.
    angulo_grados (float): Ángulo de disparo en grados (°).

    Retorna:
    (float, float): Una tupla conteniendo (altura_maxima, alcance_horizontal).
    """
    # 1. Convertir el ángulo de grados a radianes
    # ¡Como aprendimos, np.sin() y np.cos() esperan radianes!
    angulo_rad = np.deg2rad(angulo_grados)

    # 2. Calcular la altura máxima (h_max)
    # Fórmula: h_max = (v0^2 * sin^2(theta)) / (2g)
    h_max = (v0**2 * (np.sin(angulo_rad)**2)) / (2 * g)

    # 3. Calcular el alcance horizontal (R)
    # Fórmula: R = (v0^2 * sin(2*theta)) / g
    # Usamos np.sin(2 * angulo_rad)
    alcance = (v0**2 * np.sin(2 * angulo_rad)) / g

    return h_max, alcance

```

b) 	Define dos valores entre $10~m/s$ y $20~m/s$ para la rapidez inicial, y dos valores entre $25°$ y $65°$ para el ángulo de disparo. Registra los valores en la tabla. Luego usa la función escrita en el literal anterior para calcular la altura máxima y el alcance del proyectil en cada uno de los dos casos.

```python
# --- Apartado (b): Definir casos y calcular ---

# 1. Creamos la estructura de la tabla con nuestros casos de estudio.
# Elegí valores que están dentro de los rangos que pediste.
# Caso 1: 12 m/s (entre 10 y 20) y 30° (entre 25 y 65)
# Caso 2: 18 m/s (entre 10 y 20) y 50° (entre 25 y 65)
data = {
    'Caso de estudio': [1, 2],
    'Rapidez inicial [m/s]': [12.0, 18.0],
    'Ángulo de disparo [°]': [30.0, 50.0]
}

# 2. Usamos pandas para crear un DataFrame (una tabla)
df = pd.DataFrame(data)

# 3. Aplicamos nuestra función a cada fila de la tabla
# Esto es muy potente: toma v0 y ángulo de cada fila,
# llama a nuestra función y guarda los resultados en nuevas columnas.
# 'result_type='expand'' divide los resultados (h_max, alcance) en dos columnas
resultados = df.apply(
    lambda row: calcular_proyectil(row['Rapidez inicial [m/s]'], row['Ángulo de disparo [°]']),
    axis=1, # Aplica la función fila por fila
    result_type='expand'
)

# 4. Nombramos las nuevas columnas de resultados
resultados.columns = ['Altura máxima [m]', 'Alcance [m]']

# 5. Unimos la tabla original con la de resultados
df_final = pd.concat([df, resultados], axis=1)

# --- Mostrar la solución ---
print("--- Solución (a): Función ---")
print("La función 'calcular_proyectil(v0, angulo_grados)' ha sido definida.")
print(f"Usará g = {g} m/s^2.\n")

print("--- Solución (b): Tabla de Resultados ---")
# Usamos .round(3) para redondear los resultados a 3 decimales
print(df_final.round(3).to_string(index=False))
```
